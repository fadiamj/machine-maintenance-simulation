```python
import pandas as pd
import matplotlib.pyplot as plt

# Data for the scenarios
data = {
    'Scenario': ['A']*6 + ['B']*6 + ['C']*6,
    'Machine Type': ['CNC Machine', 'Conveyor Belt', 'Robotic Arm', 'Turning Machine', 'Lathe Machine', 'Hydraulic Press']*3,
    'Failure Rate': [6, 10, 10, 3, 9, 8, 5, 2, 6, 8, 7, 5, 6, 5, 8, 6, 6, 10],
    'Maintenance Time': [5, 4, 3, 5, 3, 2, 4, 2, 4, 3, 2, 2, 2, 4, 5, 2, 3, 3],
    'Number of Machines': [8, 18, 8, 11, 16, 16, 6, 12, 13, 14, 18, 5, 11, 9, 17, 14, 19, 10],
    'Maintenance Cost': [759, 991, 957, 332, 753, 933, 485, 203, 909, 578, 686, 499, 551, 980, 787, 809, 880, 305],
    'Downtime Cost': [1549, 633, 598, 522, 1865, 1368, 930, 496, 1460, 1678, 748, 651, 1806, 1051, 1086, 1621, 476, 1807],
    'Total Cost': [607.37, 649.6, 388.75, 266.88, 785.4, 4602, 1415, 82.24, 451.24, 615.27, 260.73, 383.33, 589.25, 677, 4682.5, 324, 226, 2112]
}

# Create DataFrame
df = pd.DataFrame(data)

# Visualizations
# 1. Bar plot for Failure Rate comparison across scenarios
plt.figure(figsize=(10, 6))
df.groupby(['Scenario', 'Machine Type'])['Failure Rate'].mean().unstack().plot(kind='bar', ax=plt.gca())
plt.title('Failure Rate Comparison Across Scenarios')
plt.ylabel('Failure Rate (per 1000 hours)')
plt.xlabel('Scenario')
plt.xticks(rotation=0)
plt.legend(title='Machine Type')
plt.show()

# 2. Bar plot for Maintenance Cost comparison across scenarios
plt.figure(figsize=(10, 6))
df.groupby(['Scenario', 'Machine Type'])['Maintenance Cost'].mean().unstack().plot(kind='bar', ax=plt.gca())
plt.title('Maintenance Cost Comparison Across Scenarios')
plt.ylabel('Maintenance Cost ($)')
plt.xlabel('Scenario')
plt.xticks(rotation=0)
plt.legend(title='Machine Type')
plt.show()

# 3. Bar plot for Downtime Cost comparison across scenarios
plt.figure(figsize=(10, 6))
df.groupby(['Scenario', 'Machine Type'])['Downtime Cost'].mean().unstack().plot(kind='bar', ax=plt.gca())
plt.title('Downtime Cost Comparison Across Scenarios')
plt.ylabel('Downtime Cost ($/hour)')
plt.xlabel('Scenario')
plt.xticks(rotation=0)
plt.legend(title='Machine Type')
plt.show()

# 4. Bar plot for Total Cost comparison across scenarios
plt.figure(figsize=(10, 6))
df.groupby(['Scenario', 'Machine Type'])['Total Cost'].mean().unstack().plot(kind='bar', ax=plt.gca())
plt.title('Total Cost Comparison Across Scenarios')
plt.ylabel('Total Cost ($)')
plt.xlabel('Scenario')
plt.xticks(rotation=0)
plt.legend(title='Machine Type')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Data for the three scenarios
data = {
    'Scenario': ['A', 'B', 'C'],
    'Failure Rate': [7.67, 5.5, 6.83],
    'Maintenance Time (hours)': [3.67, 2.83, 3.17],
    '# of Machines': [12.83, 11.33, 13.33],
    'Maintenance Costs': [505.17, 188.45, 498.58],
    'Downtime Costs': [711.49, 346.18, 936.55],
    'Total Cost': [1216.67, 534.63, 1435.13]
}

# Create a DataFrame
df = pd.DataFrame(data)

# Plotting the data
plt.figure(figsize=(15, 10))

# Subplot 1: Failure Rate
plt.subplot(2, 3, 1)
plt.bar(df['Scenario'], df['Failure Rate'], color=['blue', 'green', 'red'])
plt.title('Failure Rate')
plt.xlabel('Scenario')
plt.ylabel('Average')

# Subplot 2: Maintenance Time (hours)
plt.subplot(2, 3, 2)
plt.bar(df['Scenario'], df['Maintenance Time (hours)'], color=['blue', 'green', 'red'])
plt.title('Maintenance Time (hours)')
plt.xlabel('Scenario')
plt.ylabel('Average')

# Subplot 3: # of Machines
plt.subplot(2, 3, 3)
plt.bar(df['Scenario'], df['# of Machines'], color=['blue', 'green', 'red'])
plt.title('# of Machines')
plt.xlabel('Scenario')
plt.ylabel('Average')

# Subplot 4: Maintenance Costs
plt.subplot(2, 3, 4)
plt.bar(df['Scenario'], df['Maintenance Costs'], color=['blue', 'green', 'red'])
plt.title('Maintenance Costs')
plt.xlabel('Scenario')
plt.ylabel('Cost ($)')

# Subplot 5: Downtime Costs
plt.subplot(2, 3, 5)
plt.bar(df['Scenario'], df['Downtime Costs'], color=['blue', 'green', 'red'])
plt.title('Downtime Costs')
plt.xlabel('Scenario')
plt.ylabel('Cost ($)')

# Subplot 6: Total Cost
plt.subplot(2, 3, 6)
plt.bar(df['Scenario'], df['Total Cost'], color=['blue', 'green', 'red'])
plt.title('Total Cost')
plt.xlabel('Scenario')
plt.ylabel('Cost ($)')

# Adjust layout and display
plt.tight_layout()
plt.savefig("senario.png")
plt.show()
```
